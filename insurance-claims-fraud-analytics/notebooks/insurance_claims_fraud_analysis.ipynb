## Business Problem
\nAn insurer needs to identify claims that may warrant additional investigation while avoiding unnecessary reviews. The model should flag claims for human investigation, not automatically declare fraud.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('../data/processed/insurance_claims_clean.csv')
df.head()
```

## Exploratory Analysis
\nCompare review-flag rates by policy type, reporting delay, claim amount and other risk indicators.

```python
print(df['fraud_review_flag'].mean())
print(df.groupby('policy_type')['fraud_review_flag'].mean().sort_values(ascending=False))
print(df.groupby('report_delay_band')['fraud_review_flag'].mean().sort_values(ascending=False))
```

## Business Interpretation
\nThe key question is not simply whether the model predicts a flag. Management should consider investigation capacity, false positives, false negatives, customer impact, financial exposure, governance and compliance.
